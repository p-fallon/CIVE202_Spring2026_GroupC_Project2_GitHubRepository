## Global Constants
The following constants will be used throughout multiple mix design calculations:

**1 cubic yard** = 27 cubic feet

**Unit Weight of Water** = 62.4 lb/ft³

```python
# Define all constants to be used throughout the mix design

cubic_yard_ft3 = 27           # cubic ft in one cubic yard
unit_weight_water = 62.4      # unit weight of water (lb/ft³)

# Print the constants as part of an f-string
print(f"Defined Constant: 1 cubic yard = {cubic_yard_ft3} cubic ft")
print(f"Defined Constant: Unit weight of water = {unit_weight_water} lb/ft³")
```

## Function Definitions
All calculation functions are defined below in independent cells. The functions defined translate each formula from the NDOT Concrete Mix Design Excel workbook into Python. Calculations will not be performed in these cells, only defined.

```python
# Define Water Weight (Q) Calculation

# Excel Formula: Q = (A+B+C+D)*E

def calculate_water_weight_Q(weight_cement_A, weight_fly_ash_B, weight_silica_fume_C, weight_other_scm_D, ratio_water_cement_E):
    water_weight_Q = (weight_cement_A + weight_fly_ash_B + weight_silica_fume_C + weight_other_scm_D) * ratio_water_cement_E
    return water_weight_Q
```

```python
# Define Cement Volume (R) Calculation

# Excel Formula: R = A / (J * 62.4)

# The defined constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_volume_cement_R(weight_cement_A, sg_cement_J):
    volume_cement_R = weight_cement_A / (sg_cement_J * unit_weight_water)
    return volume_cement_R
```

```python
# Define Fly Ash Volume (S) Calculation

# Excel Formula: S = B / (K * 62.4)

# The defined constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_volume_fly_ash_S(weight_fly_ash_B, sg_fly_ash_K):
    volume_fly_ash_S = weight_fly_ash_B / (sg_fly_ash_K * unit_weight_water)
    return volume_fly_ash_S
```

```python
# Define Silica Fume Volume (T) Calculation

# Excel Formula: T = C / (L * 62.4)

# The defined constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_volume_silica_fume_T(weight_silica_fume_C, sg_silica_fume_L):
    volume_silica_fume_T = weight_silica_fume_C / (sg_silica_fume_L * unit_weight_water)
    return volume_silica_fume_T
```

```python
# Define Other SCM Volume (U) Calculation

# Excel Formula: U = D / (M * 62.4)

# The defined constant 'cubic_yard_ft3' (27 cubic feet per one cubic yard) to be used in place of 27 in this calculation

def calculate_volume_other_scm_U(weight_other_scm_D, sg_other_scm_M):
    volume_other_scm_U = weight_other_scm_D / (sg_other_scm_M * unit_weight_water)
    return volume_other_scm_U
```

```python
# Define Air Volume (V) Calculation

# Excel Formula: V = (F / 100) * 27

# The constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_volume_air_V(air_content_F):
    volume_air_V = (air_content_F / 100) * cubic_yard_ft3
    return volume_air_V
```

```python
# Define Water Volume (W) Calculation

# Excel Formula: W = Q / 62.4

# The constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_volume_water_W(water_weight_Q):
    volume_water_W = water_weight_Q / unit_weight_water
    return volume_water_W
```

```python
# Define Total Aggregate Volume (X) Calculation

# Excel Formula: X = 27 − R − S − T − U − V − W

# The constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_aggregate_volume_X(volume_cement_R, volume_fly_ash_S, volume_silica_fume_T, volume_other_scm_U, volume_air_V, volume_water_W):
    aggregate_volume_X = (cubic_yard_ft3 - volume_cement_R - volume_fly_ash_S - volume_silica_fume_T - volume_other_scm_U - volume_air_V - volume_water_W)
    return aggregate_volume_X
```

```python
# Define Fine Aggregate Weight (Y) Calculation

# Excel Formula: Y = 62.4 * (G / 100) * N * X

# The constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_weight_fine_aggregate_Y(percent_fine_G, sg_fine_N, aggregate_volume_X):
    weight_fine_aggregate_Y = (unit_weight_water * (percent_fine_G / 100) * sg_fine_N * aggregate_volume_X)
    return weight_fine_aggregate_Y
```

```python
# Define Coarse Aggregate Weight (Z) Calculation

# Excel Formula: Z = 62.4 * (H / 100) * O * X

# The constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_weight_coarse_aggregate_Z(percent_coarse_H, sg_coarse_O, aggregate_volume_X):
    weight_coarse_aggregate_Z = (unit_weight_water * (percent_coarse_H / 100) * sg_coarse_O * aggregate_volume_X)
    return weight_coarse_aggregate_Z
```

```python
# Define Other Aggregate Weight (AA) Calculation

# Excel Formula: AA = 62.4 * (I / 100) * P * X

# The constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_weight_other_aggregate_AA(percent_other_I, sg_other_P, aggregate_volume_X):
    weight_other_aggregate_AA = (unit_weight_water * (percent_other_I / 100) * sg_other_P * aggregate_volume_X)
    return weight_other_aggregate_AA
```

## Sequential User Inputs
The following cell will prompt users to input mix design paramaters using input(). Users will be prompted to enter each value one by one following the order from the NDOT Concrete Mix Design Excel workbook. **DO NOT INCLUDE UNITS**.

**Note:** You must enter a value for each prompt. Enter "0" for materials not present in your mix design. Aggregate Proportions (%) (G + H + I) must equal 100.

```python
# User Input Prompts

# Project Identification
project_no = int(input("Enter Project Number: "))       # Example: 404222
concrete_class = input("Enter Class of Concrete: ")     # Example: Class 47B

# Cementitious Material Weights (lb/yd³)
cement_A = float(input("Enter cement weight A (lb/yd³): "))
fly_ash_B = float(input("Enter fly ash weight B (lb/yd³): "))
silica_fume_C = float(input("Enter silica fume weight C (lb/yd³): "))
other_scm_D = float(input("Enter other SCM weight D (lb/yd³): "))

# Design Parameters
water_cement_ratio_E = float(input("Enter target water-cement ratio E: "))
air_content_F = float(input("Enter target air content F (%): "))

# Aggregate Proportions (%) - G + H + I must equal 100
percent_fine_G = float(input("Enter target percent fine aggregate G (%): "))
percent_coarse_H = float(input("Enter target percent coarse aggregate H (%): "))
percent_other_I = float(input("Enter target percent other aggregate I (%): "))

# Specific Gravities (unitless)
sg_cement_J = float(input("Enter specific gravity of cement J: "))
sg_fly_ash_K = float(input("Enter specific gravity of fly ash K: "))
sg_silica_fume_L = float(input("Enter specific gravity of silica fume L: "))
sg_other_scm_M = float(input("Enter specific gravity of other SCM M: "))
sg_fine_N = float(input("Enter specific gravity of fine aggregate N: "))
sg_coarse_O = float(input("Enter specific gravity of coarse aggregate O: "))
sg_other_P = float(input("Enter specific gravity of other aggregate P: "))

# Message to confirm all user inputs ha
print("\nAll user inputs collected successfully.")
```

## Sequential Calculations
The following cell will "call" the previously defined functions in order based on dependency. Python does not automatically resolve formula dependencies as Excel does, so it is important that these calculations are run in the correct order.

1. Water Weight (Q)
2. Cementitious Volumes (R, S, T, U)
3. Air Volume (V) & Water Volume (W)
4. Total Aggregate Volume (X)
5. Aggregate Weights (Y, Z, AA)

```python
# Calculate Water Weight (Q)

Q = calculate_water_weight_Q(cement_A, fly_ash_B, silica_fume_C, other_scm_D, water_cement_ratio_E)

print(f"Water Weight (Q): {Q:.1f} lb/yd³")
```

```python
# Calculate Cementitious Materials (R, S, T, U)

R = calculate_volume_cement_R(cement_A, sg_cement_J)
S = calculate_volume_fly_ash_S(fly_ash_B, sg_fly_ash_K)
T = calculate_volume_silica_fume_T(silica_fume_C, sg_silica_fume_L)
U = calculate_volume_other_scm_U(other_scm_D, sg_other_scm_M)

print(f"Cement volume      (R): {R:.3f} ft³")
print(f"Fly ash volume     (S): {S:.3f} ft³")
print(f"Silica fume volume (T): {T:.3f} ft³")
print(f"Other SCM volume   (U): {U:.3f} ft³")
```

```python
# Calculate Air and Water Volumes (V, W)

V = calculate_volume_air_V(air_content_F)
W = calculate_volume_water_W(Q)

print(f"Air volume   (V): {V:.3f} ft³")
print(f"Water volume (W): {W:.3f} ft³")
```

```python
# Calculate Total Aggregate Volume (X)

X = calculate_aggregate_volume_X(R, S, T, U, V, W)

print(f"Total aggregate volume (X): {X:.3f} ft³")
```

```python
# Calculate Individual Aggregate Weights (Y, Z, AA)

Y = calculate_weight_fine_aggregate_Y(percent_fine_G, sg_fine_N, X)
Z = calculate_weight_coarse_aggregate_Z(percent_coarse_H, sg_coarse_O, X)
AA = calculate_weight_other_aggregate_AA(percent_other_I, sg_other_P, X)

print(f"Fine aggregate weight   (Y):  {Y:.1f} lb/yd³")
print(f"Coarse aggregate weight (Z):  {Z:.1f} lb/yd³")
print(f"Other aggregate weight  (AA): {AA:.1f} lb/yd³")
```

```python
# Define all constants to be used throughout the mix design

cubic_yard_ft3 = 27           # cubic ft in one cubic yard
unit_weight_water = 62.4      # unit weight of water (lb/ft³)

# Print the constants as part of an f-string
print(f"Defined Constant: 1 cubic yard = {cubic_yard_ft3} cubic ft")
print(f"Defined Constant: Unit weight of water = {unit_weight_water} lb/ft³")

# Define Water Weight (Q) Calculation

# Excel Formula: Q = (A+B+C+D)*E

def calculate_water_weight_Q(weight_cement_A, weight_fly_ash_B, weight_silica_fume_C, weight_other_scm_D, ratio_water_cement_E):
    water_weight_Q = (weight_cement_A + weight_fly_ash_B + weight_silica_fume_C + weight_other_scm_D) * ratio_water_cement_E
    return water_weight_Q

# Define Cement Volume (R) Calculation

# Excel Formula: R = A / (J * 62.4)

# The defined constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_volume_cement_R(weight_cement_A, sg_cement_J):
    volume_cement_R = weight_cement_A / (sg_cement_J * unit_weight_water)
    return volume_cement_R

# Define Fly Ash Volume (S) Calculation

# Excel Formula: S = B / (K * 62.4)

# The defined constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_volume_fly_ash_S(weight_fly_ash_B, sg_fly_ash_K):
    volume_fly_ash_S = weight_fly_ash_B / (sg_fly_ash_K * unit_weight_water)
    return volume_fly_ash_S

# Define Silica Fume Volume (T) Calculation

# Excel Formula: T = C / (L * 62.4)

# The defined constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_volume_silica_fume_T(weight_silica_fume_C, sg_silica_fume_L):
    volume_silica_fume_T = weight_silica_fume_C / (sg_silica_fume_L * unit_weight_water)
    return volume_silica_fume_T

# Define Other SCM Volume (U) Calculation

# Excel Formula: U = D / (M * 62.4)

# The defined constant 'cubic_yard_ft3' (27 cubic feet per one cubic yard) to be used in place of 27 in this calculation

def calculate_volume_other_scm_U(weight_other_scm_D, sg_other_scm_M):
    volume_other_scm_U = weight_other_scm_D / (sg_other_scm_M * unit_weight_water)
    return volume_other_scm_U

# Define Air Volume (V) Calculation

# Excel Formula: V = (F / 100) * 27

# The constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_volume_air_V(air_content_F):
    volume_air_V = (air_content_F / 100) * cubic_yard_ft3
    return volume_air_V

# Define Water Volume (W) Calculation

# Excel Formula: W = Q / 62.4

# The constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_volume_water_W(water_weight_Q):
    volume_water_W = water_weight_Q / unit_weight_water
    return volume_water_W

# Define Total Aggregate Volume (X) Calculation

# Excel Formula: X = 27 − R − S − T − U − V − W

# The constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_aggregate_volume_X(volume_cement_R, volume_fly_ash_S, volume_silica_fume_T, volume_other_scm_U, volume_air_V, volume_water_W):
    aggregate_volume_X = (cubic_yard_ft3 - volume_cement_R - volume_fly_ash_S - volume_silica_fume_T - volume_other_scm_U - volume_air_V - volume_water_W)
    return aggregate_volume_X

# Define Fine Aggregate Weight (Y) Calculation

# Excel Formula: Y = 62.4 * (G / 100) * N * X

# The constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_weight_fine_aggregate_Y(percent_fine_G, sg_fine_N, aggregate_volume_X):
    weight_fine_aggregate_Y = (unit_weight_water * (percent_fine_G / 100) * sg_fine_N * aggregate_volume_X)
    return weight_fine_aggregate_Y

# Define Coarse Aggregate Weight (Z) Calculation

# Excel Formula: Z = 62.4 * (H / 100) * O * X

# The constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_weight_coarse_aggregate_Z(percent_coarse_H, sg_coarse_O, aggregate_volume_X):
    weight_coarse_aggregate_Z = (unit_weight_water * (percent_coarse_H / 100) * sg_coarse_O * aggregate_volume_X)
    return weight_coarse_aggregate_Z

# Define Other Aggregate Weight (AA) Calculation

# Excel Formula: AA = 62.4 * (I / 100) * P * X

# The constant 'unit_weight_water' (62.4 lb/ft³) to be used in place of 62.4 in this calculation

def calculate_weight_other_aggregate_AA(percent_other_I, sg_other_P, aggregate_volume_X):
    weight_other_aggregate_AA = (unit_weight_water * (percent_other_I / 100) * sg_other_P * aggregate_volume_X)
    return weight_other_aggregate_AA

# User Input Prompts

# Project Identification
project_no = int(input("Enter Project Number: "))       # Example: 404222
concrete_class = input("Enter Class of Concrete: ")     # Example: Class 47B

# Cementitious Material Weights (lb/yd³)
cement_A = float(input("Enter cement weight A (lb/yd³): "))
fly_ash_B = float(input("Enter fly ash weight B (lb/yd³): "))
silica_fume_C = float(input("Enter silica fume weight C (lb/yd³): "))
other_scm_D = float(input("Enter other SCM weight D (lb/yd³): "))

# Design Parameters
water_cement_ratio_E = float(input("Enter target water-cement ratio E: "))
air_content_F = float(input("Enter target air content F (%): "))

# Aggregate Proportions (%) - G + H + I must equal 100
percent_fine_G = float(input("Enter target percent fine aggregate G (%): "))
percent_coarse_H = float(input("Enter target percent coarse aggregate H (%): "))
percent_other_I = float(input("Enter target percent other aggregate I (%): "))

# Specific Gravities (unitless)
sg_cement_J = float(input("Enter specific gravity of cement J: "))
sg_fly_ash_K = float(input("Enter specific gravity of fly ash K: "))
sg_silica_fume_L = float(input("Enter specific gravity of silica fume L: "))
sg_other_scm_M = float(input("Enter specific gravity of other SCM M: "))
sg_fine_N = float(input("Enter specific gravity of fine aggregate N: "))
sg_coarse_O = float(input("Enter specific gravity of coarse aggregate O: "))
sg_other_P = float(input("Enter specific gravity of other aggregate P: "))

# Message to confirm all user inputs ha
print("\nAll user inputs collected successfully.")

# Calculate Water Weight (Q)

Q = calculate_water_weight_Q(cement_A, fly_ash_B, silica_fume_C, other_scm_D, water_cement_ratio_E)

print(f"Water Weight (Q): {Q:.1f} lb/yd³")

# Calculate Cementitious Materials (R, S, T, U)

R = calculate_volume_cement_R(cement_A, sg_cement_J)
S = calculate_volume_fly_ash_S(fly_ash_B, sg_fly_ash_K)
T = calculate_volume_silica_fume_T(silica_fume_C, sg_silica_fume_L)
U = calculate_volume_other_scm_U(other_scm_D, sg_other_scm_M)

print(f"Cement volume      (R): {R:.3f} ft³")
print(f"Fly ash volume     (S): {S:.3f} ft³")
print(f"Silica fume volume (T): {T:.3f} ft³")
print(f"Other SCM volume   (U): {U:.3f} ft³")

# Calculate Air and Water Volumes (V, W)

V = calculate_volume_air_V(air_content_F)
W = calculate_volume_water_W(Q)

print(f"Air volume   (V): {V:.3f} ft³")
print(f"Water volume (W): {W:.3f} ft³")

# Calculate Total Aggregate Volume (X)

X = calculate_aggregate_volume_X(R, S, T, U, V, W)

print(f"Total aggregate volume (X): {X:.3f} ft³")

# Calculate Individual Aggregate Weights (Y, Z, AA)

Y = calculate_weight_fine_aggregate_Y(percent_fine_G, sg_fine_N, X)
Z = calculate_weight_coarse_aggregate_Z(percent_coarse_H, sg_coarse_O, X)
AA = calculate_weight_other_aggregate_AA(percent_other_I, sg_other_P, X)

print(f"Fine aggregate weight   (Y):  {Y:.1f} lb/yd³")
print(f"Coarse aggregate weight (Z):  {Z:.1f} lb/yd³")
print(f"Other aggregate weight  (AA): {AA:.1f} lb/yd³")

# NDOT Concrete Mix Design Weight Summary

print("\n================================================")
print("   NDOT Concrete Mix Design — Weight Summary")
print("          (Per 1 Cubic Yard of Concrete)")
print("================================================")
print(f"  Project Number   : {project_no}")
print(f"  Concrete Class   : {concrete_class}")
print("------------------------------------------------")
print(f"    Cement (A)     : {cement_A:8.1f} lb/yd³")
print(f"    Fly Ash (B)    : {fly_ash_B:8.1f} lb/yd³")
print(f"    Silica Fume (C): {silica_fume_C:8.1f} lb/yd³")
print(f"    Other SCM (D)  : {other_scm_D:8.1f} lb/yd³")
print("------------------------------------------------")
print(f"    Fine (Y)       : {Y:8.1f} lb/yd³")
print(f"    Coarse (Z)     : {Z:8.1f} lb/yd³")
print(f"    Other (AA)     : {AA:8.1f} lb/yd³")
print("------------------------------------------------")
print(f"    Water (Q)      : {Q:8.1f} lb/yd³")
print("================================================")
print("  All weights per 1 cubic yard of concrete.")
print("================================================")
```
